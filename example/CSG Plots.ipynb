```julia
using Plots, Unitful; gr(fmt = :png)
using SolidStateDetectors
const CSG = SolidStateDetectors.ConstructiveSolidGeometry;
```

# Inverted coaxial detector

```julia
fn = "GERDA50A.yaml"
@assert isfile(fn)
sim = Simulation(fn)
```

```julia
gr(size = (500,500), grid = false, legend = false, foreground_color = :white,
    xlims = (-4u"cm", 4u"cm"), ylims = (-4u"cm",4u"cm"), zlims = (0u"cm",8u"cm"),
    xlabel = "", ylabel = "", zlabel = "", showaxis = false,
    xticks = [], yticks = [], zticks = [])
```

```julia
plot(sim.detector)
```

```julia
thecolor = :gray
plot(u"m",u"m",u"m")
plot!(CSG.EllipticalSurface(r = 0.03725), color = thecolor, linecolor = thecolor, fillalpha = 0.3)
plot!.(CSG.primitives(sim.detector.contacts[2].geometry)[2:end], color = thecolor, linecolor = thecolor, fillalpha = 0.3)
p = plot!()
```

```julia
plot(
    plot.(CSG.primitives(sim.detector.semiconductor.geometry), color = :gray, linecolor = :gray, fillalpha = 0.3)...,
    p,
    size = (2400,800), layout = (1,3)
)
#savefig("CSG_GERDA50A_semiconductor.png")
```

```julia
plot(
    plot(sim.detector, show_passives = false),
    plot.(SolidStateDetectors.ConstructiveSolidGeometry.primitives(sim.detector.contacts[1].geometry), color = 1, linecolor = 1, fillalpha = 0.3)...,
    plot.(SolidStateDetectors.ConstructiveSolidGeometry.primitives(sim.detector.contacts[2].geometry), color = 2, linecolor = 2, fillalpha = 0.3)...,
    layout = (1,8), size = (7*800,800)
)
#savefig("CSG_GERDA50A.png")
```
